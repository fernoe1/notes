Read only variables are declared using `val`
> You can't change `val` once you have given it some value.

Mutable variables are declared using `var`

To assign operators, assignment operator `=` is used.
> Variables can be declared outside main function, those variables are called top level/global variables.
>
> Keep every variable as `val` by default, change only if you need to.
