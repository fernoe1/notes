
Lists store items in the order that they were added (hence ordered list), and allow for duplicate items.

To create a read-only `List`, use the `listOf()` function.

````kotlin
val readOnlyShapes = listOf("triangle", "square", "circle")
readOnlyShapes
````

#

To create a mutable `MutableList`, use the `mutableListOf()` function.

Kotlin can infer the type of the list automatically, if you want to make this explicit, add the type of the list inside `<>` after the list declaration.

````kotlin
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
shapes
````

To prevent unwanted modifications, you can create read-only list from a mutable list by assigning it to a `List`:
```kotlin
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
val shapesLocked: List<String> = shapes
```
This is also called casting.

#

Lists are ordered, so to access an item in a list, use the indexed access operator `[]`:

````kotlin
"the first item in the list is ${readOnlyShapes[0]}"
````

#

To get the first or last item, use `.first()` and `.last()` respectively:

````kotlin
"first item in the list is ${readOnlyShapes.first()}"
````

````kotlin
"last item in the list is ${readOnlyShapes.last()}"
````

The `.first()` and `.last()` are extension functions.

#

To get the number of items in a list, use `.count()` function:

````kotlin
"this list has ${readOnlyShapes.count()} items"
````

#

To check if an item exists in a list, use `in` operator:

````kotlin
"circle" in readOnlyShapes
````

#

To add or remove items from a `MutableList`, use `.add()` and `.remove()` functions respectively:

````kotlin
shapes.add("pentagon")
shapes
````

````kotlin
shapes.remove("pentagon")
shapes
````
