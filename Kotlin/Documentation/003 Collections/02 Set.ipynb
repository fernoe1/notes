# Set
Whereas lists are ordered and allow duplicate items, sets are unordered and only store unique items.

## Read-only sets

To create read-only `Set` use the `setOf()` function.

````kotlin
val readOnlyFruit = setOf("apple", "banana", "cherry", "cherry")
readOnlyFruit
````

## Mutable sets

To create `MutableSet` use `mutableSetOf()` function.

Kotlin can infer the type of set automatically, if you want to make this explicit, pass the type of the set inside `<>`.

````kotlin
val fruits: MutableSet<String> = mutableSetOf("apple", "banana", "cherry", "cherry")
fruits
````

To prevent unwanted modifications, you can cast `MutableSet` to read-only `Set`:
```kotlin
val fruits: MutableSet<String> = mutableSEtOf("apple", "banana", "cherry", "cherry")
val fruitsLocked: Set<String> = fruits
```

> As sets are unordered, you can't access an item at a particular index.

## `.count()`

To get number of items in the set, use the `.count()` function:

````kotlin
fruits.count()
````

## `in` Operator

To check if an item is inside a set, use the `in` operator:

````kotlin
"apple" in fruits
````

## `.add()`, `.remove()`

To add or remove items from a set, use `.add()` and `.remove()` functions respectively.

````kotlin
fruits.add("dragonfruit")
fruits
````

````kotlin
fruits.remove("dragonfruit")
fruits
````
