# Type Inference

```kotlin
var customers = 10

customers = 8

customers = customers + 3
customers += 7
customers -= 3
customers *= 2
customers /= 3

customers
```

Kotlin has type inference.

`customers` is assigned an `Int` value, from this Kotlin knows that the type of `customers` is of `Int` type.

> `+=`, `-=`, `/=`, and `%=` are augmented assignment operators.
