# Conditional Expressions
Kotlin provides `if` and `when` for checking conditional expressions.
If you have to choose between `if` and `when`, `when` is recommended.

## If

To use if, add conditional expressions inside () and the code to follow if it evaluates to true in {}:

```kotlin
var d: Int
val check = true

if (check) {
    d = 1
} else {
    d = 2
}

d
```

There is no ternary operator `condition ? then : else` in Kotlin. Instead, `if` can be used as an expression. If theres only one line of action curly braces are optional:

```kotlin
val a = 1
val b = 2

if (a > b) a else b
```

## When

To use `when`:

Place the value you want to evaluate within `()`.

Place your branches within `{}`.

Use `->` in each branch to separate each check from the action to take if that check is successful.

`when` can be used as either statement or as an expression.

`when` used as statement:

```kotlin
val obj = "hello"

when (obj) {
    "1" -> "one"
    "hello" -> "greeting"
    else -> "unknown"
}
```

`when` used as an expression:

```kotlin
val result = when (obj) {
    "1" -> "one"
    "hello" -> "greeting"
    else -> "unknown"
}

result
```

`when` can be used with or without a subject.

`when` used without subject:

```kotlin
val trafficLightState = "red"

var trafficAction = when {
    trafficLightState == "green" -> "go"
    trafficLightState == "yellow" -> "slow down"
    trafficLightState == "red" -> "stop"
    else -> "malfunction"
}

trafficAction
```

`when` used with a subject:

```kotlin
trafficAction = when (trafficLightState) {
    "green" -> "go"
    "yellow" -> "slow down"
    "red" -> "stop"
    else -> "malfunction"
}

trafficAction
```

Using `when` with subject makes your code easier to read and maintain.

# Ranges

## `..` Operator

`1..4` is equivalent to `1, 2, 3, 4`

## `.<` Operator

`1.<4` is equivalent to `1, 2, 3`

## `downTo`

`4 downTo 1` is equivalent to `4, 3, 2, 1`

## `step`

`1..5 step 2` is equivalent to `1, 3, 5`

`'a' .. 'd'` is equivalent to `'a', 'b', 'c', 'd'`

`'z' downTo 's' step 2` is equivalent to `'z', 'x', 'v', 't'`

# Loops
Use for to iterate over a range, use while to do something till something happens.

## For

Place iterator and range within `()` with keyword `in`, action you want to do inside `{}`:

```kotlin
for (number in 1..5) print(number)
// number is iterator, 1..5 is range
```

Collections can also be iterated over:

```kotlin
val cakes = listOf("carrot", "chocolate", "strawberry")
for (cake in cakes) {
    println("Yummy $cake cake")
}
```

## While

While can be used in two ways.

To execute a code block when the condition is true `while`.


To execute a code block first then check the conditional expression `do-while`.

First use-case:

```kotlin
var cakesEaten = 0
while (cakesEaten < 3) {
    println("Cakes eated $cakesEaten")
    cakesEaten++
}
```

Second use-case:

```kotlin
var cakesBaked = 0
do {
    println("Cakes baked $cakesBaked")
    cakesBaked++
} while (cakesBaked < cakesEaten)
```
