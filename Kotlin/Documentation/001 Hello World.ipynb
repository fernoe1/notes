# Hello, world!

```kotlin
fun main() {
    println("Hello, world!")
}

main()
```

`fun` is used to declare a function.

`main()` function is where your program starts from.

Body of function is writed within `{}`

`println()` and `print()` write to stdout, one with linebreak one without.

The `main()` function is being called for notebook purposes, in real program, you don't call it.

# Variables

```kotlin
val popcorn = 5
val hotdog = 7
var customers = 10
customers = 8
customers
```

Read only variables are declared using `val`
> You can't change `val` once you have given it some value.

Mutable variables are declared using `var`

To assign operators, assignment operator `=` is used.
> Variables can be declared outside main function, those variables are called top level/global variables.
>
> Keep every variable as `val` by default, change only if you need to.


# String Templates

You can use template expressions to access data stored in variables and objects, and convert them to strings.

To do so you need to use the dollar sign `$` inside a string literal `"`

To do template expression use curly braces in conjunction with a `$` sign, e.g. `${}`

```kotlin
val customers = 10
```

```kotlin
"there are $customers customers"
```

```kotlin
"there are ${customers + 1} customers"
```
