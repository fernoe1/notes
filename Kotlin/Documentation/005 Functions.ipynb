# Functions

````kotlin
fun hello() {
    return println("Hello, world!")
}

hello()
````

In Kotlin:

Function parameters are written within `()`.

Each parameter must have a type, and separated by a comma if you have multiple.

Return type of function is written after `()` using `:`.

The body of a function is written within `{}`.

The `return` keyword is used to either exit or return something from a function.

````kotlin
fun sum(x: Int, y: Int):Int {
    return x + y
}

sum(1, 2)
````

It is recommended you name your function using camelCase, with no underscores.

## Named arguments

When calling functions you don't need to include parameter names, but if you do, you can write them in any order.

````kotlin
fun printMessageWithPrefix(message: String, prefix: String) {
    println("[$prefix] $message")
}

printMessageWithPrefix(prefix = "Log", message = "Hello")
````

To define default value use `=` after the type.

````kotlin
fun printMessageWithPrefix(message: String, prefix: String = "Info") {
    println("[$prefix] $message")
}

printMessageWithPrefix("Hello")
````

When skipping default values, every other parameter after that must be passed using named arguments.

## Functions without return

If your function doesn't return a useful value, then it's return type is `Unit`. Kotlin automatically sets return type of those function to `Unit`, so you don't need to write it explicitly.

## Single expression functions

````kotlin
fun sum(x: Int, y: Int) = x + y

sum(1, 2)
````

When you use `=` operator to create functions, you can omit `return` and `{}`. Also since you are using `=` operator, Kotlin uses type inference, so you can omit the return type as well.

> If you use `{}` to declare your function's body, you must declare `return` type unless its `Unit` type.

# Lambda expressions

The following code utilizes lambda expression:

````kotlin
val upperCaseString = { text: String -> text.uppercase() }
upperCaseString("hello")
````

Lambda expressions are written within `{}`.

The parameters followed by `->`.

The function body after `->`.

> If you declare lambda without parameters, there's no need for `->`.
> ```kotlin
> { println("message") }
> ```

## Pass to another function

````kotlin
val numbers = listOf(1, -2, 3, -4, 5)
val positiveNumbers = numbers.filter { x -> x > 0 }
positiveNumbers
````

````kotlin
val isNegative = { x: Int -> x < 0 }
val negativeNumbers = numbers.filter(isNegative)
negativeNumbers
````

> If the lambda expression is the only parameter, you can omit the `()` as shown in the first use-case.

````kotlin
val doubled = numbers.map { x -> x * 2 }
doubled
````

## Function types

In Kotlin, functions also have types, and Kotlin's type inference can infer function's type based on its parameters, but sometimes you might need to explicitly set it.

The syntax for function type:

Each parameters type written in `()` separated by `,`.

The return type written after `->`.

````kotlin
val upperCaseString: (String) -> String = { text -> text.uppercase() }
upperCaseString("hello")
````

If your lambda function has no parameters, then `() -> Unit`.

> You must declare function type either in lambda expression or in the function type, otherwise your code won't compile.
> The following code won't work: `val upperCaseString = { str -> str.uppercase() }`

## Return from a function

````kotlin
fun toSeconds(time: String): (Int) -> Int = when (time) {
    "hour" -> { value -> value * 60 * 60 }
    "minute" -> { value -> value * 60 }
    "second" -> { value -> value }
    else -> { value -> value }
}

val timesInMinutes = listOf(2, 10, 15, 1)
val min2sec = toSeconds("minute")
val totalTimeInSeconds = timesInMinutes.map(min2sec).sum()
"Total time is $totalTimeInSeconds secs"
````

## Invoke separately

Lambda expressions can be invoked on their own by passing required arguments inside `()` after `{}`.

````kotlin
{ text: String -> text.uppercase() }("hello")
````

## Trailing lambdas

If lambda expression is the only parameter `()` can be omitted, and if it is the last parameter, it can be passed outside `()`. Such syntax is called trailing lambda.

````kotlin
(1..3).fold(0, { x, item -> x + item })
````

````kotlin
(1..3).fold(0) {x, item -> x + item }
````
