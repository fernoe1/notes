# List
Lists store items in the order that they were added (hence ordered list), and allow for duplicate items.

## Read only lists

To create a read-only `List`, use the `listOf()` function.

````kotlin
val readOnlyShapes = listOf("triangle", "square", "circle")
readOnlyShapes
````

## Mutable lists

To create a mutable `MutableList`, use the `mutableListOf()` function.

Kotlin can infer the type of the list automatically, if you want to make this explicit, add the type of the list inside `<>` after the list declaration.

````kotlin
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
shapes
````

To prevent unwanted modifications, you can create read-only list from a mutable list by assigning it to a `List`:
```kotlin
val shapes: MutableList<String> = mutableListOf("triangle", "square", "circle")
val shapesLocked: List<String> = shapes
```
This is also called casting.

## `[]` Operator

Lists are ordered, so to access an item in a list, use the indexed access operator `[]`:

````kotlin
"the first item in the list is ${readOnlyShapes[0]}"
````

## `.first()`, `.last()`

To get the first or last item, use `.first()` and `.last()` respectively:

````kotlin
"first item in the list is ${readOnlyShapes.first()}"
````

````kotlin
"last item in the list is ${readOnlyShapes.last()}"
````

The `.first()` and `.last()` are extension functions.

## `.count()`

To get the number of items in a list, use `.count()` function:

````kotlin
"this list has ${readOnlyShapes.count()} items"
````

## `in` Operator

To check if an item exists in a list, use `in` operator:

````kotlin
"circle" in readOnlyShapes
````

## `.add()`, `.remove()`

To add or remove items from a `MutableList`, use `.add()` and `.remove()` functions respectively:

````kotlin
shapes.add("pentagon")
shapes
````

````kotlin
shapes.remove("pentagon")
shapes
````

# Set
Whereas lists are ordered and allow duplicate items, sets are unordered and only store unique items.

## Read-only sets

To create read-only `Set` use the `setOf()` function.

````kotlin
val readOnlyFruit = setOf("apple", "banana", "cherry", "cherry")
readOnlyFruit
````

## Mutable sets

To create `MutableSet` use `mutableSetOf()` function.

Kotlin can infer the type of set automatically, if you want to make this explicit, pass the type of the set inside `<>`.

````kotlin
val fruits: MutableSet<String> = mutableSetOf("apple", "banana", "cherry", "cherry")
fruits
````

To prevent unwanted modifications, you can cast `MutableSet` to read-only `Set`:
```kotlin
val fruits: MutableSet<String> = mutableSEtOf("apple", "banana", "cherry", "cherry")
val fruitsLocked: Set<String> = fruits
```

> As sets are unordered, you can't access an item at a particular index.

## `.count()`

To get number of items in the set, use the `.count()` function:

````kotlin
fruits.count()
````

## `in` Operator

To check if an item is inside a set, use the `in` operator:

````kotlin
"apple" in fruits
````

## `.add()`, `.remove()`

To add or remove items from a set, use `.add()` and `.remove()` functions respectively.

````kotlin
fruits.add("dragonfruit")
fruits
````

````kotlin
fruits.remove("dragonfruit")
fruits
````

# Map
Maps store items as key-value pairs.
> In a map every key must be unique.
>
> Values can be duplicated.

## Read-only maps

To create read-only `Map` use `mapOf()` function.

The easiest way to create a map, is to use `to` between each key and its respective value.

````kotlin
val readOnlyJuiceMenu = mapOf("apple" to 100, "kiwi" to 190, "orange" to 100)
readOnlyJuiceMenu
````

## Mutable maps

To create `MutableMap` use `mutableMapOf()` function.

When creating maps Kotlin can infer the type of the keys and values automatically, to do this explicitly, add the types of keys and values inside `<>`.

````kotlin
val juiceMenu: MutableMap<String, Int> = mutableMapOf("apple" to 100, "kiwi" to 190, "orange" to 100)
juiceMenu
````

To prevent unwanted modifications, you can cast `MutableMap` to read-only `Map`:
```kotlin
val juiceMenu: MutableMap<String, Int> = mutableMapOf("apple" to 100, "kiwi" to 190, "orange" to 100)
val juiceMenuLocked: Map<String, Int> = juiceMenu
```

## `[]` Operator

To access value in the map, use `[]` operator with its key:

````kotlin
juiceMenu["apple"]
````

If you try to access key-value pair in a map with a key that doesn't exist, you will get `null` value:

````kotlin
juiceMenu["pineapple"]
````

You can also use indexed access operator (`[]` operator) to add key-value pair to a `MutableMap`:

````kotlin
juiceMenu["coconut"] = 150
juiceMenu
````

## `.remove()`

To remove key-value pairs from a `MutableMap` use `.remove()` function:

````kotlin
juiceMenu.remove("coconut")
juiceMenu
````

## `.count()`

To get count of key-value pairs, use the `.count()` function:

````kotlin
juiceMenu.count()
````

## `.containsKey()`

To check if specific key exists inside a map, use `.containsKey()` function:

````kotlin
juiceMenu.contains("apple")
````

## `.keys`, `.values`

To obtain collection of keys or values of a map, use `keys` and `values` properties respectively:

````kotlin
juiceMenu.keys
````

````kotlin
juiceMenu.values
````

`keys` and `values` are examples of properties of an object, to access them we use the `.` notation.

To check if a key or value exists in a map, you can use keys and values properties in conjunction with `in` operator:

````kotlin
"apple" in juiceMenu.keys
````

Alternatively, you can omit `.keys` property:

````kotlin
"apple" in juiceMenu
````

````kotlin
200 in juiceMenu.values
````
